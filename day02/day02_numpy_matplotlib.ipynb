## 1. Creating Arrays

An array is a grid of numbers, all the same type. Think of it like a matrix (or a multi-dimensional table).

```python
# From a Python list
a = np.array([1, 2, 3, 4, 5])
print("1D array:", a)
print("Shape:", a.shape)   # (5,) — a 1D array with 5 elements
print("Dtype:", a.dtype)   # int64 — NumPy chose integer type

# 2D array (matrix)
b = np.array([[1, 2, 3],
              [4, 5, 6]])
print("\n2D array:\n", b)
print("Shape:", b.shape)   # (2, 3) — 2 rows, 3 columns
```

```python
# Common creation functions — you'll use these constantly
print("zeros(3, 4):\n", np.zeros((3, 4)))         # All zeros
print("\nones(2, 3):\n", np.ones((2, 3)))          # All ones
print("\narange(0, 10, 2):", np.arange(0, 10, 2))  # Like range(), but returns array
print("\nlinspace(0, 1, 5):", np.linspace(0, 1, 5))  # 5 evenly-spaced points from 0 to 1
print("\neye(3):\n", np.eye(3))                    # Identity matrix
```

## 2. Shapes, Reshaping & Dimensions

In deep learning, **shape** is everything. You'll constantly check and manipulate shapes.

Think of shape as dimensions:
- `(5,)` → 1D: a vector with 5 elements
- `(3, 4)` → 2D: a matrix with 3 rows × 4 columns
- `(2, 3, 4)` → 3D: 2 "slices" of 3×4 matrices
- `(B, C, H, W)` → 4D: batch of images (batch_size, channels, height, width)

```python
# Reshape — change the shape without changing the data
a = np.arange(12)  # [0, 1, 2, ..., 11]
print("Original:", a, "shape:", a.shape)

b = a.reshape(3, 4)  # Reshape to 3 rows × 4 columns
print("\nReshaped (3, 4):\n", b)

c = a.reshape(2, 2, 3)  # Reshape to 3D
print("\nReshaped (2, 2, 3):\n", c)

# -1 means "figure it out automatically"
d = a.reshape(4, -1)  # 4 rows, auto-compute columns → (4, 3)
print("\nReshaped (4, -1):\n", d, "shape:", d.shape)
```

```python
# Adding/removing dimensions — common in DL
x = np.array([1, 2, 3])  # shape: (3,)
print("Original shape:", x.shape)

# Add a dimension
row = x[np.newaxis, :]   # shape: (1, 3) — row vector
col = x[:, np.newaxis]   # shape: (3, 1) — column vector
print("Row vector shape:", row.shape)
print("Col vector shape:", col.shape)

# Flatten — collapse all dimensions to 1D
matrix = np.array([[1, 2], [3, 4], [5, 6]])  # (3, 2)
flat = matrix.flatten()  # → (6,)
print("\nFlattened:", flat, "shape:", flat.shape)
```

## 3. Indexing & Slicing

```python
m = np.array([[10, 20, 30],
              [40, 50, 60],
              [70, 80, 90]])

print("Full matrix:\n", m)
print("\nElement [1, 2]:", m[1, 2])        # Row 1, Col 2 → 60
print("Row 0:", m[0])                       # First row → [10, 20, 30]
print("Column 1:", m[:, 1])                 # All rows, column 1 → [20, 50, 80]
print("Submatrix [0:2, 1:3]:\n", m[0:2, 1:3])  # Top-right 2×2 block

# Boolean indexing — select elements that meet a condition
print("\nElements > 50:", m[m > 50])
```

## 4. Broadcasting

Broadcasting lets NumPy do math on arrays with *different shapes* by automatically expanding the smaller one.

**Rule:** dimensions are compared from right to left. A dimension is compatible if:
1. They're equal, or
2. One of them is 1 (it gets "stretched")

```python
# Scalar + array: scalar is broadcast to match array shape
a = np.array([1, 2, 3])
print("a + 10 =", a + 10)     # [11, 12, 13]
print("a * 2 =", a * 2)       # [2, 4, 6]

# Matrix + vector: vector is broadcast across rows
matrix = np.array([[1, 2, 3],
                   [4, 5, 6]])     # shape: (2, 3)
vector = np.array([10, 20, 30])    # shape: (3,)

result = matrix + vector           # vector is added to each row
print("\nMatrix + Vector:\n", result)

# This is how normalization works in DL:
# subtract mean and divide by std for each feature
data = np.array([[100, 0.5],
                 [200, 0.8],
                 [150, 0.6]])
mean = data.mean(axis=0)  # Mean of each column
std = data.std(axis=0)    # Std of each column
normalized = (data - mean) / std  # Broadcasting!
print("\nNormalized:\n", normalized)
```

## 5. Vectorized Math (No Loops!)

NumPy operations work on entire arrays at once — this is **much faster** than Python loops.

In DL, you should almost never write loops over individual elements.

```python
# Element-wise operations
a = np.array([1.0, 2.0, 3.0, 4.0])
b = np.array([5.0, 6.0, 7.0, 8.0])

print("a + b =", a + b)
print("a * b =", a * b)      # Element-wise multiply
print("a ** 2 =", a ** 2)    # Square each element
print("np.sqrt(a) =", np.sqrt(a))
print("np.exp(a) =", np.exp(a))
print("np.sum(a) =", np.sum(a))
print("np.mean(a) =", np.mean(a))

# Dot product — essential for neural networks!
print("\na · b (dot product) =", np.dot(a, b))  # 1*5 + 2*6 + 3*7 + 4*8 = 70

# Matrix multiplication
W = np.array([[1, 2], [3, 4], [5, 6]])  # (3, 2)
x = np.array([0.5, 1.5])                # (2,)
result = W @ x  # Matrix-vector multiply: (3, 2) @ (2,) → (3,)
print("\nW @ x =", result)
```

```python
# Speed comparison: vectorized vs. loop
import time

size = 1_000_000
a = np.random.randn(size)
b = np.random.randn(size)

# Loop version
start = time.time()
result_loop = sum(a[i] * b[i] for i in range(size))
time_loop = time.time() - start

# Vectorized version
start = time.time()
result_vec = np.dot(a, b)
time_vec = time.time() - start

print(f"Loop:       {time_loop:.4f}s")
print(f"Vectorized: {time_vec:.6f}s")
print(f"Speedup:    {time_loop / time_vec:.0f}×")
```

## 6. Random Numbers

Random numbers are used everywhere in DL: weight initialization, data shuffling, dropout, noise, etc.

```python
# Set seed for reproducibility (same random numbers every time)
np.random.seed(42)

# Uniform random numbers [0, 1)
print("Uniform:", np.random.rand(5))

# Normal (Gaussian) random numbers — mean=0, std=1
print("Normal:", np.random.randn(5))

# Random integers
print("Integers:", np.random.randint(0, 10, size=5))

# Random matrix (e.g., simulating a weight matrix)
W = np.random.randn(3, 4) * 0.01  # Small random values (like weight init)
print("\nWeight matrix (3×4):\n", W)
```

## 7. Plotting with Matplotlib

Visualization is essential for understanding what's happening during training.

```python
# ── Line plot: loss over epochs ────────────────────────
epochs = np.arange(1, 51)
train_loss = 2.5 * np.exp(-0.05 * epochs) + 0.1 * np.random.randn(50) * 0.1
val_loss = 2.5 * np.exp(-0.04 * epochs) + 0.2 + 0.1 * np.random.randn(50) * 0.1

plt.figure(figsize=(8, 4))
plt.plot(epochs, train_loss, label="Train Loss", linewidth=2)
plt.plot(epochs, val_loss, label="Val Loss", linewidth=2, linestyle="--")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Training vs Validation Loss")
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# ── Scatter plot: 2D data ─────────────────────────────
np.random.seed(42)
class_a = np.random.randn(50, 2) + np.array([2, 2])
class_b = np.random.randn(50, 2) + np.array([-2, -2])

plt.figure(figsize=(6, 6))
plt.scatter(class_a[:, 0], class_a[:, 1], label="Class A", alpha=0.7)
plt.scatter(class_b[:, 0], class_b[:, 1], label="Class B", alpha=0.7)
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.title("Two Classes of Data Points")
plt.legend()
plt.grid(True, alpha=0.3)
plt.axis("equal")
plt.tight_layout()
plt.show()
```

```python
# ── Subplots: multiple plots side by side ─────────────
fig, axes = plt.subplots(1, 3, figsize=(14, 4))

x = np.linspace(-3, 3, 100)

# Plot 1: Functions
axes[0].plot(x, np.sin(x), label="sin(x)")
axes[0].plot(x, np.cos(x), label="cos(x)")
axes[0].set_title("Trig Functions")
axes[0].legend()
axes[0].grid(True, alpha=0.3)

# Plot 2: Histogram
data = np.random.randn(1000)
axes[1].hist(data, bins=30, alpha=0.7, color="steelblue", edgecolor="white")
axes[1].set_title("Normal Distribution")
axes[1].grid(True, alpha=0.3)

# Plot 3: Heatmap (like a 2D feature map)
matrix = np.random.randn(8, 8)
im = axes[2].imshow(matrix, cmap="viridis")
axes[2].set_title("Heatmap (Feature Map)")
plt.colorbar(im, ax=axes[2])

plt.tight_layout()
plt.show()
```

## 8. NumPy ↔ PyTorch Cheat Sheet

Here's a preview of how NumPy maps to PyTorch (you'll use this starting Day 4):

| NumPy | PyTorch | Notes |
|-------|---------|-------|
| `np.array([1,2,3])` | `torch.tensor([1,2,3])` | |
| `np.zeros((3,4))` | `torch.zeros(3,4)` | No tuple needed in PyTorch |
| `np.random.randn(3,4)` | `torch.randn(3,4)` | |
| `a.shape` | `a.shape` | Same! |
| `a.reshape(2,3)` | `a.reshape(2,3)` or `a.view(2,3)` | |
| `a.dtype` | `a.dtype` | Same! |
| `np.dot(a, b)` | `torch.dot(a, b)` | |
| `a @ b` | `a @ b` | Same! |
| `a.sum()` | `a.sum()` | Same! |
| `np_array` | `torch.from_numpy(np_array)` | Convert |
| — | `tensor.numpy()` | Convert back |

## 📝 Key Takeaways

1. **Arrays have shapes** — always check `.shape`. In DL, shape mismatches are the #1 bug.
2. **Vectorize everything** — use array operations, not loops. It's 100x+ faster.
3. **Broadcasting** — NumPy/PyTorch auto-expand dimensions when doing math. Learn the rules.
4. **Reshaping is free** — `.reshape()` doesn't copy data, just changes how you view it.
5. **Matplotlib basics** — `plt.plot`, `plt.scatter`, `plt.imshow`, `plt.subplots` will cover 90% of your plotting needs.
