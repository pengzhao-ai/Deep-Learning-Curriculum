## 1. Variables, Types & Basic Operations

In Python, you don't declare types — you just assign values. Python figures out the type automatically.

```python
# Variables — no need to declare types
x = 10            # int
y = 3.14          # float
name = "PyTorch"  # str (string)
is_training = True # bool (boolean)

# Check the type of a variable
print(type(x))        # <class 'int'>
print(type(y))        # <class 'float'>
print(type(name))     # <class 'str'>
print(type(is_training))  # <class 'bool'>
```

```python
# Basic math operations
a = 10
b = 3

print("a + b =", a + b)     # Addition: 13
print("a - b =", a - b)     # Subtraction: 7
print("a * b =", a * b)     # Multiplication: 30
print("a / b =", a / b)     # Division (always float): 3.333...
print("a // b =", a // b)   # Integer division: 3
print("a % b =", a % b)     # Modulo (remainder): 1
print("a ** b =", a ** b)   # Exponent: 1000
```

## 2. Lists, Tuples & Dicts

These are the three most common *containers* in Python.

| Container | Syntax | Mutable? | Use case |
|-----------|--------|----------|----------|
| **List** | `[1, 2, 3]` | ✅ Yes | Ordered collection you can change |
| **Tuple** | `(1, 2, 3)` | ❌ No | Fixed collection (e.g., shapes) |
| **Dict** | `{"key": value}` | ✅ Yes | Key-value mapping (e.g., config) |

```python
# ── Lists ──────────────────────────────────────────────
losses = [2.5, 1.8, 1.2, 0.9, 0.5]

print("First loss:", losses[0])       # Indexing starts at 0
print("Last loss:", losses[-1])       # Negative index = from the end
print("First 3:", losses[:3])         # Slicing: [start:stop] (stop is exclusive)
print("Length:", len(losses))          # Number of elements

# Add to list
losses.append(0.3)
print("After append:", losses)
```

```python
# ── Tuples ─────────────────────────────────────────────
# In DL, you'll see tuples for shapes: (batch_size, channels, height, width)
image_shape = (32, 3, 224, 224)

print("Batch size:", image_shape[0])
print("Channels:", image_shape[1])

# Unpacking — assign each element to a variable
batch, channels, height, width = image_shape
print(f"Image: {channels}ch × {height}×{width}, batch={batch}")
```

```python
# ── Dicts ──────────────────────────────────────────────
# You'll use dicts for hyperparameters, configs, results
config = {
    "learning_rate": 0.001,
    "epochs": 50,
    "batch_size": 32,
    "optimizer": "Adam",
}

print("LR:", config["learning_rate"])
print("All keys:", list(config.keys()))
print("All values:", list(config.values()))

# Add a new key
config["model"] = "CNN"
print("Updated config:", config)
```

## 3. Loops & Conditionals

```python
# ── for loop ───────────────────────────────────────────
# The most common loop in DL code
losses = [2.5, 1.8, 1.2, 0.9, 0.5]

for loss in losses:
    print(f"Loss: {loss:.2f}")

print("---")

# range() generates numbers: range(5) → 0, 1, 2, 3, 4
for epoch in range(5):
    print(f"Epoch {epoch + 1}")

print("---")

# enumerate() gives you both index AND value — very common!
for i, loss in enumerate(losses):
    print(f"  Step {i}: loss = {loss}")
```

```python
# ── if / elif / else ───────────────────────────────────
accuracy = 0.85

if accuracy > 0.9:
    print("Great model!")
elif accuracy > 0.7:
    print("Decent model, keep training.")
else:
    print("Model needs work.")
```

## 4. Functions

Functions let you reuse code. In DL, you'll write functions for training loops, evaluation, plotting, etc.

```python
# Basic function
def compute_mse(predictions, targets):
    """Compute Mean Squared Error."""
    total = 0
    for pred, target in zip(predictions, targets):
        total += (pred - target) ** 2
    return total / len(predictions)

preds = [2.5, 0.0, 2.1]
targets = [3.0, -0.5, 2.0]
mse = compute_mse(preds, targets)
print(f"MSE = {mse:.4f}")
```

```python
# Default arguments — very common in DL
def train(model_name, epochs=10, lr=0.001):
    """Simulate training."""
    print(f"Training {model_name} for {epochs} epochs with lr={lr}")

train("CNN")                    # Uses defaults: epochs=10, lr=0.001
train("ViT", epochs=50)         # Override epochs
train("MLP", lr=0.01, epochs=20)  # Override both (order doesn't matter with named args)
```

## 5. f-strings (Formatted Strings)

f-strings let you embed variables and expressions inside strings. You'll use them constantly for printing training progress.

```python
epoch = 15
loss = 0.3847
accuracy = 0.9213

# Basic f-string
print(f"Epoch {epoch}: loss = {loss}, accuracy = {accuracy}")

# Format specifiers
print(f"Epoch {epoch:03d}: loss = {loss:.4f}, accuracy = {accuracy:.2%}")
#       03d = pad with zeros, 3 digits wide
#       .4f = 4 decimal places
#       .2% = percentage with 2 decimal places
```

## 6. List Comprehensions

A compact way to create lists. You'll see these everywhere in Python/DL code.

```python
# Traditional loop
squares_loop = []
for x in range(10):
    squares_loop.append(x ** 2)

# Same thing as a list comprehension — one line!
squares_comp = [x ** 2 for x in range(10)]

print("Loop:", squares_loop)
print("Comp:", squares_comp)

# With a condition (filter)
even_squares = [x ** 2 for x in range(10) if x % 2 == 0]
print("Even squares:", even_squares)
```

## 7. Classes — The Foundation of PyTorch Models

This is the **most important section** for DL. Every PyTorch model is a Python class.

A class is a blueprint that bundles **data** (attributes) and **behavior** (methods) together.

Key concepts:
- `__init__` — called when you *create* an object (constructor)
- `self` — refers to the current object
- Inheritance — one class can *extend* another (this is how `nn.Module` works)

```python
# ── Basic class ────────────────────────────────────────
class Neuron:
    def __init__(self, weight, bias):
        """Constructor — runs when you create a Neuron."""
        self.weight = weight  # Store as attribute
        self.bias = bias
    
    def forward(self, x):
        """Compute output: y = weight * x + bias."""
        return self.weight * x + self.bias
    
    def __repr__(self):
        """How the object looks when printed."""
        return f"Neuron(w={self.weight}, b={self.bias})"

# Create an instance
n = Neuron(weight=2.0, bias=1.0)
print(n)                    # Neuron(w=2.0, b=1.0)
print(n.forward(3.0))       # 2.0 * 3.0 + 1.0 = 7.0
print(n.weight)             # Access attribute: 2.0
```

```python
# ── Inheritance ────────────────────────────────────────
# In PyTorch, every model inherits from nn.Module.
# Here's a simplified version to understand the pattern.

class BaseModel:
    """A simplified base class (like nn.Module)."""
    def __init__(self):
        self.layers = []
    
    def count_params(self):
        return len(self.layers)


class MyModel(BaseModel):  # MyModel inherits from BaseModel
    def __init__(self):
        super().__init__()  # Call parent's __init__  ← You'll see this in EVERY PyTorch model
        self.layers = ["conv1", "conv2", "fc1"]
    
    def forward(self, x):
        print(f"Processing through {len(self.layers)} layers")
        return x


model = MyModel()
print(f"Number of layers: {model.count_params()}")  # Inherited method
model.forward("input_data")  # Own method
```

```python
# ── __call__ — making objects callable ─────────────────
# In PyTorch, you call a model like a function: model(x)
# This works because nn.Module defines __call__ which invokes forward().

class CallableNeuron:
    def __init__(self, weight, bias):
        self.weight = weight
        self.bias = bias
    
    def __call__(self, x):
        """This makes the object callable like a function."""
        return self.weight * x + self.bias

neuron = CallableNeuron(2.0, 1.0)

# Now you can call it like a function!
result = neuron(3.0)  # Same as neuron.__call__(3.0)
print(f"neuron(3.0) = {result}")  # 7.0

# This is exactly how PyTorch models work:
# model(x)  →  calls model.__call__(x)  →  which calls model.forward(x)
```

## 8. Importing Modules

Python organizes code into modules (files) and packages (folders). Here are the import patterns you'll see in DL code:

```python
# Common import patterns in DL
import math                        # Import entire module
print(math.sqrt(16))               # Use as math.sqrt()

from math import pi, exp           # Import specific items
print(pi)                          # Use directly, no prefix
print(exp(1))                      # e^1 ≈ 2.718

import numpy as np                 # Import with alias (nickname)
print(np.array([1, 2, 3]))         # Use as np.xxx

# You'll write these in almost every notebook:
# import torch
# import torch.nn as nn
# import torch.optim as optim
# import matplotlib.pyplot as plt
# from torch.utils.data import DataLoader, Dataset
```

## 9. Useful Built-in Functions

A few Python built-ins you'll encounter frequently:

```python
numbers = [3, 1, 4, 1, 5, 9, 2, 6]

print("len:", len(numbers))       # Length
print("sum:", sum(numbers))       # Sum
print("min:", min(numbers))       # Minimum
print("max:", max(numbers))       # Maximum
print("sorted:", sorted(numbers)) # Sorted copy

# zip — pair up two lists (used a lot in training loops)
names = ["train_loss", "val_loss", "accuracy"]
values = [0.5, 0.7, 0.92]

for name, value in zip(names, values):
    print(f"  {name}: {value}")

# isinstance — check type (useful for debugging)
print("\nis 42 an int?", isinstance(42, int))
print("is 3.14 an int?", isinstance(3.14, int))
```
